```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
engine= create_engine("mysql+mysqlconnector://root:Sivesh2308@localhost/sales_analysis")
```

```python
query = """
SELECT
    o.invoice_no,
    o.invoice_date,
    c.customer_id,
    c.country,
    oi.quantity,
    oi.unit_price
FROM orders o
JOIN order_items oi ON o.invoice_no = oi.invoice_no
JOIN customers c ON o.customer_id = c.customer_id
"""
df = pd.read_sql(query, engine)
df.head()

```

```python
df['invoice_date']= pd.to_datetime(df['invoice_date'])
```

```python
df=df[df['quantity']>0]
df=df[df['unit_price']>0]
```

```python
df['revenue'] = df['quantity'] * df['unit_price']
```

```python
total_revenue_all = df['revenue'].sum()
```

```python
clean_df = df.dropna(subset=['customer_id', 'invoice_no'])
```

```python
total_revenue_clean = clean_df['revenue'].sum()
```
